```python
#Q5. Find the number of missing values in every column. Display the columns in descending order of missing values.
df= df.isnull().sum().sort_values(ascending=False)
print(df)
```

```python
#Q6. Check whether the dataset contains duplicate rows. If duplicates exist, remove them and verify the result.
print(df.duplicated().sum())
df = df.drop_duplicates()
print(df.duplicated().sum())
print(len(df))
```

```python
#Q7. Convert Order_Date into the appropriate datetime datatype. Then find the earliest and latest order dates in the dataset.
df["Order_Date"] = pd.to_datetime(
    df["Order_Date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

print(df["Order_Date"].dtype)
print(df["Order_Date"].min())
print(df["Order_Date"].max())
```

```python
#Q8. Find all orders where Quantity is greater than 3, Rating is less than or equal to 2, and Order_Status is 'Delivered'. How many such orders are there?
print(len(df[
    (df["Quantity"] > 3) & (df["Rating"] <= 2) & (df["Order_Status"] == "Delivered")
]))
```

```python
#Q9. Display the top 10 orders based on Sales. Show only Order_ID, Customer_Name, Product, Quantity, and Sales.
print(df.nlargest(10, "Sales")[[ "Order_ID", "Customer_Name", "Product", "Quantity", "Sales"]])
```

```python
#Q10. Find all orders where the discount is greater than 20%. Sort the result by Discount from highest to lowest.
df = df[df["Discount"] > 20].sort_values( "Discount", ascending=False)
print(df)
```

```python
#Q11. Find the 10 products with the highest average rating. Display the product name and average rating.
products = df.groupby("Product")["Rating"].mean().sort_values(
    ascending=False
).head(10)

print(products)
```

```python
#Q12. Calculate the total sales generated by each Category. Sort the result from highest to lowest sales.
print(
    df.groupby("Category")["Sales"]
      .sum()
      .sort_values(ascending=False)
)
```

```python
#Q13. Find the total quantity sold and total sales for each Product. Display the top 10 products based on total sales.
result = df.groupby("Product").agg(
    Total_Quantity=("Quantity", "sum"),
    Total_Sales=("Sales", "sum")
)
print(result.sort_values("Total_Sales", ascending=False).head(10))
```

```python
#Q14. Calculate the average Sales, average Rating, and average Delivery_Days for each Payment_Mode.
result = df.groupby("Payment_Mode").agg(
    Average_Sales=("Sales", "mean"),
    Average_Rating=("Rating", "mean"),
    Average_Delivery_Days=("Delivery_Days", "mean")
)
print(result)
```

```python
#Q15. Find the number of orders and total sales generated by each City. Which 5 cities have generated the highest sales?
df = df.groupby("City").agg(
    Number_of_Orders=("Order_ID", "count"),
    Total_Sales=("Sales", "sum")
)
print(df)
```

```python
#Q16. Create a new column called Month from Order_Date. Find the total sales for each month.
df["Order_Date"] = pd.to_datetime(df["Order_Date"], format="mixed")
df["Month"] = df["Order_Date"].dt.to_period("M")
print(df.groupby("Month")["Sales"].sum())
```

```python
#Q18. Compare the performance of different categories by calculating Total Sales, Average Sales, Total Quantity Sold, Average Rating, and Average Delivery Days. Which category appears to perform best overall?
df = df.groupby("Category").agg(
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Average_Rating=("Rating", "mean"),
    Average_Delivery_Days=("Delivery_Days", "mean")
)
print(df)
```

```python
#Q19. Create a new column called Calculated_Sales using: Quantity × Unit_Price × (1 − Discount). Compare it with the existing Sales column and find the rows where the difference is greater than 1.
df["Calculated_Sales"] = (
    df["Quantity"] *
    df["Unit_Price"] *
    (1 - df["Discount"])
)

df["Difference"] = abs(
    df["Calculated_Sales"] - df["Sales"]
)

result = df[df["Difference"] > 1]

print(result)
```

```python
#Q20. Create a new column called Customer_Segment based on Customer_Age: Age < 25 → Young; 25–40 → Adult; 41–60 → Middle Age; Above 60 → Senior. Then calculate the number of orders, total sales, and average rating for each customer segment.
df["Customer_Segment"] = pd.cut(
    df["Customer_Age"],
    bins=[0, 24, 40, 60, float("inf")],
    labels=["Young", "Adult", "Middle Age", "Senior"]
)

result = df.groupby("Customer_Segment", observed=True).agg(
    Number_of_Orders=("Order_ID", "count"),
    Total_Sales=("Sales", "sum"),
    Average_Rating=("Rating", "mean")
)

print(result)
```
